## 2. Target or Proxy

My proposed target is whether a page is **declining**.

I would create a binary target from `trend_direction`:

- `down` → 1 (declining)
- `up` or `flat` → 0 (not declining)

This target is a proxy for identifying pages that may deserve closer review. It does not mean that every declining page should automatically be refreshed, because a decline does not prove that a refresh will improve performance.

The model would use observable signals available before the decision, such as content age, days since last update, impressions, average position, CTR, and word count.

## 3. Success Metric

## 5. Why ML Beats a Fixed Rule

A fixed rule such as "stale and visible pages should be reviewed first" is transparent and useful as a baseline, but it treats the chosen thresholds as fixed.

ML can consider several observable signals together, such as content age, update history, impressions, average position, CTR, and word count. This may allow it to identify combinations of signals that a simple rule would miss.

The purpose is not to assume that ML will always outperform a human-designed rule. I will compare the model against the baseline and only keep the additional complexity if it produces useful improvement.

The final output should support human review rather than automatically deciding which pages must be refreshed.
